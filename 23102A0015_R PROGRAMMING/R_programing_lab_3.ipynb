```python
%%R

# Basics of Calculations

2 + 3
10 - 4
5 * 6
20 / 4
```

```python
%%R

# Power Operations

2^3
2^0.5
2^-0.5

c(2,3,5,7)^2
```

```python
#Integer Division

%%R

2 %/% 2
5 %/% 2
7 %/% 3

c(2,3,5,7) %/% 2
```

```python
# Modulo Division
%%R

2 %% 2
3 %% 2
7 %% 3
7 %% 4

c(2,3,5,7) %% 2
```

```python
%%R

# Maximum
max(c(1.2, 3.4, -7.8))

```

```python
%%R
# Minimum
min(c(1.2, 3.4, -7.8))
```

```python
%%R
# Absolute value
abs(c(-1,-2,-3,4,5))

# Square root
sqrt(c(4,9,16,25))

# Sum
sum(c(2,3,5,7))

# Product
prod(c(2,3,5,7))

# Round
round(1.83)

# Natural logarithm
log(10)

# Log base 10
log10(100)
```

```python
%%R
# row-wise Matrix
y = matrix(
  nrow=4,
  ncol=2,
  data=c(1,2,3,4,5,6,7,8),
  byrow=TRUE
)

y
```

```python
%%R
# Filter rows where the first column of matrix y is greater than 2
filtered_y <- y[y[,1] > 2,]
display(filtered_y)
```
